```python
online_pass=40
online_fail=10
offline_pass=25
offline_fail=25
online_total=online_pass+online_fail
offline_total=offline_pass+offline_fail
pass_total=online_pass+offline_pass
fail_total=online_fail+offline_fail
total=online_total+offline_total
E1 = (online_total * pass_total) /total
E2 = (online_total * fail_total) / total
E3 = (offline_total * pass_total) / total
E4 = (offline_total * fail_total) / total
chi_square = ((online_pass - E1) ** 2 / E1 +
              (online_fail - E2) ** 2 / E2 +
              (offline_pass - E3) ** 2 / E3 +
              (offline_fail - E4) ** 2 / E4)
alpha=0.05
critical_value = 3.841
print("Expected values:")
print(E1, E2, E3, E4)

print("Chi-square =", chi_square)
print("Alpha =", alpha)
print("Critical value =", critical_value)


if chi_square > critical_value:
    print("Thus Reject H0")
else:
    print("Thus Accept H0")


```

```python
x = [10, 30, 20, 40, 50]
y = [8, 6, 4, 2, 10]


x_mean = sum(x) / len(x)
y_mean = sum(y) / len(y)

print("X Mean =", x_mean)
print("Y Mean =", y_mean)

xy = []
x2 = []
y2 = []

for i in range(len(x)):
    a = x[i] - x_mean
    b = y[i] - y_mean

    xy.append(a * b)
    x2.append(a ** 2)
    y2.append(b ** 2)

print("XY =", xy)
print("X2 =", x2)
print("Y2 =", y2)

sum_xy = sum(xy)
sum_x2 = sum(x2)
sum_y2 = sum(y2)

print("Sum XY =", sum_xy)
print("Sum X2 =", sum_x2)
print("Sum Y2 =", sum_y2)

r = sum_xy / ((sum_x2 * sum_y2) ** 0.5)

print("Karl Pearson r =", r)
```
