```python
import matplotlib.pyplot as plt
```

```python
#IMPLEMENT BAR CHART
subjects = ["Math", "Science","English","Computer"]
marks = [85,90,78,95]
plt.bar(subjects, marks, color="orange")
plt.title("Student Marks")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.show()
```

```python
#IMPLEMENT LINE CHART
import matplotlib.pyplot as plt
month = ["jan","feb","mar","apr","may"]
sales =[20,40,30,60,10]
plt.plot(month,sales,marker="o",color="red")
plt.title("Monthly Sales Report")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

```python
#IMPLEMENT PIE CHART
import matplotlib.pyplot as plt
household = ["Rent","Food","Education","Others"]
expenses=[40,30,20,10]
plt.pie(expenses,labels=household,autopct="%1.1f%%",startangle=90)
plt.title("Household Expenses")
plt.show()
```

```python
#IMPLEMENT HISTOGRAM
import matplotlib.pyplot as plt
marks=[5,8,12,18,19,22,25,28,35,38,42,45,50,55,58,62,68,72,78,85,92]
plt.hist(marks,bins=[0,20,40,60,80,100],color="skyblue",edgecolor="black")
plt.title("Students'Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.show()
```

```python
#IMPLEMENT SCATTER PLOT
import matplotlib.pyplot as plt
height=[150,155,160,165,170,175]
weight=[45,50,55,60,65,70]
plt.scatter(height,weight,color="red")
plt.title("Height vs Weight")
plt.xlabel("Height (cm)")
plt.ylabel("Weight (kg)")
plt.show()
```

```python
#IMPLEMENT BOX PLOT
import matplotlib.pyplot as plt
marks=[45,50,55,60,65,70,72,75,80,85,90,95]
plt.boxplot(marks)
plt.title("Plots of Student Marks")
plt.ylabel("Marks")
plt.show()
```

```python
# IMPLEMENT HEAT MAP
import matplotlib.pyplot as plt

marks = [
    [85, 90, 78, 95],
    [75, 80, 88, 92],
    [90, 85, 95, 89],
    [70, 75, 82, 80]
]

plt.imshow(marks, cmap="YlOrRd")

plt.colorbar(label="Marks")
plt.title("Student Marks Heat Map")

plt.xticks([0, 1, 2, 3], ["Math", "Science", "English", "Computer"])
plt.yticks([0, 1, 2, 3], ["Student 1", "Student 2", "Student 3", "Student 4"])

plt.show()
```
