```python
#two sample equal variance:

import math

A = [6, 7, 8, 9, 10]
B = [8, 9, 10, 11, 12]

n1 = len(A)
n2 = len(B)

# Mean
mean1 = sum(A) / n1
mean2 = sum(B) / n2

print("Mean A =", mean1)
print("Mean B =", mean2)

# S1^2
s1_2 = sum((x - mean1)**2 for x in A) / (n1 - 1)
print("S1^2 =", s1_2)

# S2^2
s2_2 = sum((x - mean2)**2 for x in B) / (n2 - 1)
print("S2^2 =", s2_2)

# Pooled variance
sp2 = ((n1 - 1)*s1_2 + (n2 - 1)*s2_2) / (n1 + n2 - 2)
print("Pooled Variance (Sp^2) =", sp2)

# Standard Error
SE = math.sqrt(sp2 * (1/n1 + 1/n2))
print("Standard Error =", SE)

# t statistic
t = (mean1 - mean2) / SE
print("t statistic =", t)

# Degree of freedom
df = n1 + n2 - 2
print("df =", df)

# Decision
t_critical = 2.306

if abs(t) > t_critical:
    print("Reject H0")
else:
    print("Fail to reject H0")
```

```python
# one sample z test mean:

import math

x = [4, 5, 6, 5, 7, 3]

# Hypothetical mean
mu = 5

# 1. Mean
n = len(x)
mean = sum(x) / n
print("Mean =", mean)

# 2. Variance S^2
s2 = sum((i - mean)**2 for i in x) / (n - 1)
print("S^2 =", s2)

# 3. Standard deviation
s = math.sqrt(s2)
print("Standard Deviation =", s)

# 4. t-statistic
t = (mean - mu) / (s / math.sqrt(n))
print("t-statistic =", t)

# 5. Degree of freedom
df = n - 1
print("df =", df)

# 6. Decision
t_critical = 2.571

if abs(t) > t_critical:
    print("Reject H0")
else:
    print("Fail to reject H0")
```

```python
import math

n1 = 36
mean1 = 52
sd1 = 6

n2 = 36
mean2 = 49
sd2 = 6

# Standard Error
SE = math.sqrt((sd1**2 / n1) + (sd2**2 / n2))
print("SE =", SE)

# Z statistic
z = (mean1 - mean2) / SE
print("Z =", z)

# 5% level
z_critical = 1.96

if abs(z) > z_critical:
    print("Reject H0")
else:
    print("Accept H0")
```
