```python
# COMPARISON OF ITERATION 1 AND ITERATION 2

import numpy as np

x = np.array([1, 2, 3])
y = np.array([2, 4, 5])

m = 0.01
w = 0
b = 0
n = len(x)

def cost(w, b):
    y_pred = w * x + b
    return (1/n) * sum((y - y_pred) ** 2)

results = []

for i in range(2):

    y_pred = w * x + b

    dw = (-2/n) * sum(x * (y - y_pred))
    db = (-2/n) * sum(y - y_pred)

    w = w - m * dw
    b = b - m * db

    y_pred = w * x + b
    j = cost(w, b)

    results.append([i + 1, w, b, j, y_pred])

# Comparison
print("ITERATION COMPARISON")
print("-" * 60)
print("Iteration |     w     |     b     |   Cost")
print("-" * 60)

for r in results:
    print(f"{r[0]:9} | {r[1]:.4f} | {r[2]:.4f} | {r[3]:.4f}")

print("\nPrediction Comparison:")
for r in results:
    print(f"Iteration {r[0]}: {r[4]}")

print("\nCost decreased:",
      results[0][3], "->", results[1][3])
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Data
x = np.array([1, 2, 3])
y = np.array([2, 4, 5])

# Learning rate
m = 0.01

w = 0
b = 0
n = len(x)

def cost(w, b):
    y_pred = w * x + b
    return (1/n) * sum((y - y_pred) ** 2)

# Store values for comparison
iterations = []
costs = []
weights = []
biases = []

for i in range(2):

    # Prediction
    y_pred = w * x + b

    # Gradients
    dw = (-2/n) * sum(x * (y - y_pred))
    db = (-2/n) * sum(y - y_pred)

    # Update
    w = w - m * dw
    b = b - m * db

    # New prediction and cost
    y_pred = w * x + b
    j = cost(w, b)

    iterations.append(i + 1)
    costs.append(j)
    weights.append(w)
    biases.append(b)

    print("Iteration:", i + 1)
    print("w:", w)
    print("b:", b)
    print("Cost:", j)
    print("y_pred:", y_pred)
    print("-" * 40)


# ---------------- GRAPH 1: COST COMPARISON ----------------

plt.figure(figsize=(6, 4))
plt.plot(iterations, costs, marker='o')
plt.xlabel("Iteration")
plt.ylabel("Cost")
plt.title("Cost vs Iteration")
plt.xticks(iterations)
plt.grid(True)
plt.show()


# ---------------- GRAPH 2: ACTUAL vs PREDICTED ----------------

plt.figure(figsize=(6, 4))

plt.scatter(x, y, color="blue", label="Actual Data")

# Predictions for iteration 1
y_pred_1 = weights[0] * x + biases[0]

# Predictions for iteration 2
y_pred_2 = weights[1] * x + biases[1]

plt.plot(x, y_pred_1, color="red", label="Iteration 1")
plt.plot(x, y_pred_2, color="green", label="Iteration 2")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Actual vs Predicted Values")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# THW WHOLE CODE :

import numpy as np
import matplotlib.pyplot as plt

x = np.array([1, 2, 3])
y = np.array([2, 4, 5])

m = 0.01
w = 0
b = 0
n = len(x)

def cost(w, b):
    y_pred = w * x + b
    return (1/n) * sum((y - y_pred) ** 2)

# Store values for comparison
iteration_values = []

for i in range(2):

    # Prediction
    y_pred = w * x + b

    # Gradients
    dw = (-2/n) * sum(x * (y - y_pred))
    db = (-2/n) * sum(y - y_pred)

    # Updation
    w = w - m * dw
    b = b - m * db

    # Substitute the value in y_pred
    y_pred = w * x + b

    # Update the cost
    j = cost(w, b)

    # Store values
    iteration_values.append((i + 1, w, b, j, y_pred.copy()))

    print("the iteration: ", i + 1)
    print("the new cost: ", j)
    print("the new y_pred is: ", y_pred)
    print("the new w is: ", w)
    print("the new b is: ", b)
    print("the new cost is: ", cost(w, b))
    print("-" * 40)


# ---------------- COMPARISON ----------------

print("\nCOMPARISON OF ITERATION 1 AND ITERATION 2")
print("-" * 50)

for data in iteration_values:
    i, w_value, b_value, cost_value, prediction = data

    print("Iteration:", i)
    print("w:", w_value)
    print("b:", b_value)
    print("Cost:", cost_value)
    print("y_pred:", prediction)
    print("-" * 50)


# ---------------- SCATTER GRAPH ----------------

iteration1 = iteration_values[0]
iteration2 = iteration_values[1]

y_pred_1 = iteration1[4]
y_pred_2 = iteration2[4]

plt.figure(figsize=(8, 5))

# Actual data
plt.scatter(x, y, color="blue", s=80, label="Actual Data")

# Iteration 1
plt.scatter(
    x, y_pred_1,
    color="red",
    s=80,
    label="Iteration 1 Prediction"
)

# Iteration 2
plt.scatter(
    x, y_pred_2,
    color="green",
    s=80,
    label="Iteration 2 Prediction"
)

# Connect predictions
plt.plot(x, y_pred_1, color="red", linestyle="--")
plt.plot(x, y_pred_2, color="green", linestyle="--")

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Scatter Graph: Iteration 1 vs Iteration 2")
plt.legend()
plt.grid(True)

plt.show()
```

```python
import numpy as np

# Given data
x = np.array([1, 2, 3])
y = np.array([2, 4, 5])

# Cost function: J(theta) = theta^3
def J(theta):
    return theta ** 3

# Gradient: dJ/dtheta = 3*theta^2
def gradient(theta):
    return 3 * theta ** 2

# Gradient Descent parameters
theta = 1.0       # initial theta
alpha = 0.1       # learning rate
iterations = 3

print("Gradient Descent for J(theta) = theta^3")
print("-" * 55)

for i in range(iterations + 1):
    cost = J(theta)
    grad = gradient(theta)

    print(f"Iteration {i}:")
    print(f"  theta    = {theta:.6f}")
    print(f"  J(theta) = {cost:.6f}")
    print(f"  gradient = {grad:.6f}")

    # Update theta
    theta = theta - alpha * grad
```

```python
import matplotlib.pyplot as plt

# Cost function: J(theta) = theta^3
def cost(theta):
    return theta ** 3

# Gradient: dJ/dtheta = 3(theta^2)
def gradient(theta):
    return 3 * theta ** 2

# Initial values
theta = 2
learning_rate = 0.01
iterations = 3

# Store values for graph
iteration_values = []
cost_values = []

# Gradient Descent
for i in range(iterations + 1):
    iteration_values.append(i)
    cost_values.append(cost(theta))

    print("Iteration:", i)
    print("Theta:", theta)
    print("Gradient:", gradient(theta))
    print("Cost:", cost(theta))
    print("----------------------")

    # Update theta
    theta = theta - learning_rate * gradient(theta)

# Cost vs Iteration graph
plt.plot(iteration_values, cost_values, marker='o')

plt.xlabel("Iteration")
plt.ylabel("Cost J(theta)")
plt.title("Cost vs Iteration")
plt.grid(True)

plt.show()
```
