```python
import math

A=[6,7,8,9,10]
B=[8,9,10,11,12]

m1=len(A)
m2=len(B)

mean1=sum(A)/m1
mean2=sum(B)/m2

print("Mean A =",mean1)
print("Mean B =",mean2)

s1_2=sum((x-mean1)**2 for x in A)/(m1-1)
print("S1^2 =",s1_2)

s2_2=sum((x-mean2)**2 for x in B)/(m2-1)
print("S2^2 =",s2_2)

sp2=((m1-1)*s1_2+(m2-1)*s2_2)/(m1+m2-2)

print("Pooled Variance (Sp^2) =",sp2)

SE=math.sqrt(sp2*(1/m1+1/m2))

print("Standard Error =",SE)

t=(mean1-mean2)/SE

print("t-statistic =",t)

df=m1+m2-2

print("Degrees of Freedom =",df)

t_critical=2.306

print("t-critical =",t_critical)

if abs(t)>t_critical:
    print("Reject H0")
else:
    print("Fail to Reject H0")
```

```python
import math

x=[4,5,6,7,8,9]

mu=5

sigma=float(input("Enter population S.D: "))

n=len(x)

mean=sum(x)/n

print("Sample Mean =",mean)

SE=sigma/math.sqrt(n)

print("Standard Error =",SE)

z=(mean-mu)/SE

print("Z-statistic =",z)

z_critical=1.96

print("Z-critical =",z_critical)

if abs(z)>z_critical:
    print("Reject H0")
else:
    print("Fail to Reject H0")
```
