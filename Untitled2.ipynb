```python
import numpy as np

A=np.array([9,11,13,9,8])
B=np.array([13,12,10,15,5])
C=np.array([14,13,11,7,9])

mean_a=np.mean(A)
mean_b=np.mean(B)
mean_c=np.mean(C)

total=np.sum(A)+np.sum(B)+np.sum(C)

N=len(A)+len(B)+len(C)

grand_mean=total/N

n=5

SSC=n*((mean_a-grand_mean)**2+(mean_b-grand_mean)**2+(mean_c-grand_mean)**2)

print("SSC =",SSC)

k=3
df1=k-1

print("df1 =",df1)

MS_between=SSC/df1

print("MS between =",MS_between)

SSE_A=np.sum((A-mean_a)**2)
SSE_B=np.sum((B-mean_b)**2)
SSE_C=np.sum((C-mean_c)**2)

SSE=SSE_A+SSE_B+SSE_C

print("SSE =",SSE)

df2=N-k

print("df2 =",df2)

MS_within=SSE/df2

print("MS within =",MS_within)

F_calculated=MS_between/MS_within

print("F calculated =",F_calculated)

F_tabulated=3.89

print("F tabulated =",F_tabulated)

if F_calculated>F_tabulated:
    print("Reject H0")
else:
    print("Do not reject H0")
```

```python
import pandas as pd

student=['A','B','C','D','E']

x=[80,45,55,60,70]
y=[85,40,50,65,75]

df=pd.DataFrame({
    'Student':student,
    'Maths (X)':x,
    'Rx':pd.Series(x).rank(ascending=True),
    'Statistics (Y)':y,
    'Ry':pd.Series(y).rank(ascending=True)
})

df['d']=df['Rx']-df['Ry']
df['d2']=df['d']**2

sum_d2=df['d2'].sum()
n=len(df)

rs=1-(6*sum_d2)/(n*(n**2-1))

print(df)
print("Sum of d2 =",sum_d2)
print("Spearman Rank Correlation =",rs)

if rs>0:
    print("Positive Relation")
elif rs<0:
    print("Negative Relation")
else:
    print("No Relation")
```
